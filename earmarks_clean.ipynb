## One risk number per town
following AdaptaBrasil lables

```python
df_final['Índice Risco Médio'] = (df_final['Índice Deslizamento'] + df_final['Índice Inundação'])/2

def classificar(valor):
    if pd.isna(valor):
        return 'No data'
    if valor <= 0.19:
        return 'Very low'
    if valor <= 0.39:
        return 'Low'
    if valor <= 0.59:
        return 'Medium'
    if valor <= 0.79:
        return 'High'
    return 'Very high'

df_final['Label Risco Médio'] = df_final['Índice Risco Médio'].apply(classificar)

df_final['Label Risco Médio'].value_counts()
```

## Events vs money by state

```python
#mapping states to initials
ufs = {'ACRE': 'AC', 'ALAGOAS': 'AL', 'AMAPÁ': 'AP', 'AMAZONAS': 'AM', 'BAHIA': 'BA', 'CEARÁ': 'CE', 'DISTRITO FEDERAL': 'DF', 'ESPÍRITO SANTO': 'ES', 'GOIÁS': 'GO', 'MARANHÃO': 'MA', 'MATO GROSSO': 'MT', 'MATO GROSSO DO SUL': 'MS', 'MINAS GERAIS': 'MG', 'PARÁ': 'PA', 'PARAÍBA': 'PB', 'PARANÁ': 'PR', 'PERNAMBUCO': 'PE', 'PIAUÍ': 'PI', 'RIO DE JANEIRO': 'RJ', 'RIO GRANDE DO NORTE': 'RN', 'RIO GRANDE DO SUL': 'RS', 'RONDÔNIA': 'RO', 'RORAIMA': 'RR', 'SANTA CATARINA': 'SC', 'SÃO PAULO': 'SP', 'SERGIPE': 'SE', 'TOCANTINS': 'TO'}

eventos_uf = recentes.groupby('Sigla_UF').size().reset_index(name='Eventos')

dfambientais['Sigla_UF'] = dfambientais['UF'].map(ufs)
dinheiro_uf = dfambientais.groupby('Sigla_UF')['Valor Empenhado'].sum().reset_index()

estados = eventos_uf.merge(dinheiro_uf, on='Sigla_UF', how='left')
estados['Valor Empenhado'] = estados['Valor Empenhado'].fillna(0)

estados['% Eventos'] = (estados['Eventos'] / estados['Eventos'].sum() * 100).round(1)
estados['% Dinheiro'] = (estados['Valor Empenhado'] / estados['Valor Empenhado'].sum() * 100).round(1)

estados.sort_values('% Eventos', ascending=False)
```

```python
CSV ESTADOS EVENTOS DINHEIRO
```

```python
estados.to_csv('data/estados_eventos_dinheiro.csv', index=False, sep=';', encoding='utf-8-sig', decimal=',')
```

```python
#Slope file for Flourish
slope = estados[['Sigla_UF']].copy()
slope['Share of events'] = estados['Eventos'] / estados['Eventos'].sum()*100
slope['Share of money'] = estados['Valor Empenhado'] / estados['Valor Empenhado'].sum() * 100

slope.to_csv('data/slope_estados.csv', index=False)

slope.head()
```
